Import pandas

```python
import pandas as pd
```

Create a DataFrame with balance.txt

```python
balance_df = pd.read_csv("balance.txt", sep=" ")
print(balance_df)
```

Compare the average income based on ethnicity.

```python
print("The average income of African Americans is", balance_df[balance_df.Ethnicity == "African American"].loc[:,"Income"].mean())
print("The average income of Asians is", balance_df[balance_df.Ethnicity == "Asian"].loc[:,"Income"].mean())
print("The average income of Caucasians is", balance_df[balance_df.Ethnicity == "Caucasian"].loc[:,"Income"].mean())
```

On average, do married or single people have a higher balance?

```python
print("The average balance for married people is", balance_df[balance_df.Gender == "Female"].loc[:,"Balance"].mean())
print("The average balance for single people is", balance_df[balance_df.Gender == "Male"].loc[:,"Balance"].mean())
print("Therefore, single people have a higher balance on average")
```

What is the highest income in our dataset?

```python
print("The highest income in the dataset is", balance_df["Income"].max())
```

What is the lowest income in our dataset?

```python
print("The lowest income in the dataset is", balance_df["Income"].min())
```

How many cards do we have recorded in our dataset? (Hint: use `sum()`)

```python
print("The number of cards recorded in the dataset is", balance_df["Cards"].sum())
```


How many females do we have information for vs how many males? (Hint: use `count())

```python
print("The number of females in the dataset is", balance_df[balance_df.Gender == "Female"].loc[:,"Balance"].count())
print("The number of males in the dataset is", balance_df[balance_df.Gender == "Male"].loc[:,"Balance"].count())
```
